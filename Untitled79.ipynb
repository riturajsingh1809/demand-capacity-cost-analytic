```python
import pandas as pd
import numpy as np

# ============================================================
# 1. LOAD CLEANED DATASET
# ============================================================

file_name = "Cleaned_Demand_Capacity_Cost_Analytics.xlsx"

df = pd.read_excel(file_name)

print("=" * 60)
print("PYTHON ANALYSIS RESULT")
print("=" * 60)

print("\nDataset Shape:", df.shape)

print("\nFirst 5 Rows:")
display(df.head())


# ============================================================
# 2. BASIC DATA INFORMATION
# ============================================================

print("\n" + "=" * 60)
print("1. BASIC DATA INFORMATION")
print("=" * 60)

print("\nNumber of Rows:", df.shape[0])
print("Number of Columns:", df.shape[1])

print("\nData Types:")
display(df.dtypes.to_frame("Data Type"))

print("\nMissing Values:")
display(df.isnull().sum().to_frame("Missing Values"))

print("\nDuplicate Rows:", df.duplicated().sum())


# ============================================================
# 3. OVERALL BUSINESS KPIs
# ============================================================

total_records = len(df)

total_demand = df["demand_units"].sum()

total_capacity = df["available_capacity"].sum()

total_projected_demand = df["projected_demand"].sum()

total_projected_capacity = df["projected_capacity"].sum()

total_infrastructure_cost = df["infrastructure_cost"].sum()

total_operating_cost = df["operating_cost"].sum()

total_cost = (
    total_infrastructure_cost +
    total_operating_cost
)

overall_utilization = (
    total_demand /
    total_capacity * 100
)

projected_utilization = (
    total_projected_demand /
    total_projected_capacity * 100
)

capacity_gap = (
    total_capacity -
    total_demand
)

projected_gap = (
    total_projected_capacity -
    total_projected_demand
)

print("\n" + "=" * 60)
print("2. OVERALL BUSINESS KPIs")
print("=" * 60)

print("Total Records:", total_records)
print("Total Demand Units:", round(total_demand, 2))
print("Total Available Capacity:", round(total_capacity, 2))
print("Overall Utilization: {:.2f}%".format(overall_utilization))
print("Total Projected Demand:", round(total_projected_demand, 2))
print("Total Projected Capacity:", round(total_projected_capacity, 2))
print("Projected Utilization: {:.2f}%".format(projected_utilization))
print("Total Infrastructure Cost:", round(total_infrastructure_cost, 2))
print("Total Operating Cost:", round(total_operating_cost, 2))
print("Total Cost:", round(total_cost, 2))
print("Current Capacity Gap:", round(capacity_gap, 2))
print("Projected Capacity Gap:", round(projected_gap, 2))


# ============================================================
# 4. REGION ANALYSIS
# ============================================================

region_summary = (
    df.groupby("region")
    .agg(
        demand_units=("demand_units", "sum"),
        available_capacity=("available_capacity", "sum"),
        projected_demand=("projected_demand", "sum"),
        projected_capacity=("projected_capacity", "sum"),
        infrastructure_cost=("infrastructure_cost", "sum"),
        operating_cost=("operating_cost", "sum"),
        total_cost=("total_cost", "sum"),
        incidents=("incidents", "sum")
    )
    .reset_index()
)

region_summary["utilization_rate"] = (
    region_summary["demand_units"] /
    region_summary["available_capacity"] * 100
)

region_summary["projected_utilization_rate"] = (
    region_summary["projected_demand"] /
    region_summary["projected_capacity"] * 100
)

region_summary["capacity_gap"] = (
    region_summary["available_capacity"] -
    region_summary["demand_units"]
)

region_summary["projected_gap"] = (
    region_summary["projected_capacity"] -
    region_summary["projected_demand"]
)

print("\n" + "=" * 60)
print("3. REGION ANALYSIS")
print("=" * 60)

display(
    region_summary.sort_values(
        "demand_units",
        ascending=False
    )
)


# ============================================================
# 5. BUSINESS SEGMENT ANALYSIS
# ============================================================

segment_summary = (
    df.groupby("business_segment")
    .agg(
        demand_units=("demand_units", "sum"),
        available_capacity=("available_capacity", "sum"),
        projected_demand=("projected_demand", "sum"),
        projected_capacity=("projected_capacity", "sum"),
        total_cost=("total_cost", "sum"),
        incidents=("incidents", "sum")
    )
    .reset_index()
)

segment_summary["utilization_rate"] = (
    segment_summary["demand_units"] /
    segment_summary["available_capacity"] * 100
)

segment_summary["capacity_gap"] = (
    segment_summary["available_capacity"] -
    segment_summary["demand_units"]
)

print("\n" + "=" * 60)
print("4. BUSINESS SEGMENT ANALYSIS")
print("=" * 60)

display(
    segment_summary.sort_values(
        "demand_units",
        ascending=False
    )
)


# ============================================================
# 6. SERVICE TYPE ANALYSIS
# ============================================================

service_summary = (
    df.groupby("service_type")
    .agg(
        demand_units=("demand_units", "sum"),
        available_capacity=("available_capacity", "sum"),
        total_cost=("total_cost", "sum"),
        downtime_hours=("downtime_hours", "sum"),
        incidents=("incidents", "sum")
    )
    .reset_index()
)

service_summary["utilization_rate"] = (
    service_summary["demand_units"] /
    service_summary["available_capacity"] * 100
)

service_summary["cost_per_capacity_unit"] = (
    service_summary["total_cost"] /
    service_summary["available_capacity"]
)

print("\n" + "=" * 60)
print("5. SERVICE TYPE ANALYSIS")
print("=" * 60)

display(
    service_summary.sort_values(
        "total_cost",
        ascending=False
    )
)


# ============================================================
# 7. MONTHLY TREND ANALYSIS
# ============================================================

df["month"] = pd.to_datetime(
    df["month"],
    errors="coerce"
)

monthly_summary = (
    df.groupby("month")
    .agg(
        demand_units=("demand_units", "sum"),
        available_capacity=("available_capacity", "sum"),
        projected_demand=("projected_demand", "sum"),
        projected_capacity=("projected_capacity", "sum"),
        total_cost=("total_cost", "sum")
    )
    .reset_index()
    .sort_values("month")
)

monthly_summary["utilization_rate"] = (
    monthly_summary["demand_units"] /
    monthly_summary["available_capacity"] * 100
)

monthly_summary["capacity_gap"] = (
    monthly_summary["available_capacity"] -
    monthly_summary["demand_units"]
)

print("\n" + "=" * 60)
print("6. MONTHLY TREND ANALYSIS")
print("=" * 60)

display(monthly_summary)


# ============================================================
# 8. CAPACITY SHORTAGE ANALYSIS
# ============================================================

shortage_df = df[
    df["capacity_gap"] < 0
].copy()

print("\n" + "=" * 60)
print("7. CAPACITY SHORTAGE ANALYSIS")
print("=" * 60)

print(
    "Shortage Records:",
    len(shortage_df)
)

print(
    "Total Shortage Units:",
    round(
        abs(shortage_df["capacity_gap"].sum()),
        2
    )
)

if len(shortage_df) > 0:

    shortage_summary = (
        shortage_df
        .groupby(["region", "service_type"])
        .agg(
            shortage_units=(
                "capacity_gap",
                lambda x: abs(x.sum())
            ),
            shortage_records=(
                "capacity_gap",
                "count"
            )
        )
        .reset_index()
        .sort_values(
            "shortage_units",
            ascending=False
        )
    )

    display(shortage_summary)

else:

    print("No current capacity shortage found.")


# ============================================================
# 9. PROJECTED CAPACITY SHORTAGE
# ============================================================

projected_shortage_df = df[
    df["projected_gap"] < 0
].copy()

print("\n" + "=" * 60)
print("8. PROJECTED CAPACITY SHORTAGE")
print("=" * 60)

print(
    "Projected Shortage Records:",
    len(projected_shortage_df)
)

print(
    "Total Projected Shortage Units:",
    round(
        abs(
            projected_shortage_df[
                "projected_gap"
            ].sum()
        ),
        2
    )
)

if len(projected_shortage_df) > 0:

    projected_shortage_summary = (
        projected_shortage_df
        .groupby("region")
        .agg(
            projected_shortage_units=(
                "projected_gap",
                lambda x: abs(x.sum())
            ),
            records=(
                "projected_gap",
                "count"
            )
        )
        .reset_index()
        .sort_values(
            "projected_shortage_units",
            ascending=False
        )
    )

    display(projected_shortage_summary)

else:

    print("No projected capacity shortage found.")


# ============================================================
# 10. COST ANALYSIS
# ============================================================

cost_analysis = (
    df.groupby("region")
    .agg(
        infrastructure_cost=(
            "infrastructure_cost",
            "sum"
        ),
        operating_cost=(
            "operating_cost",
            "sum"
        ),
        total_cost=(
            "total_cost",
            "sum"
        ),
        available_capacity=(
            "available_capacity",
            "sum"
        ),
        demand_units=(
            "demand_units",
            "sum"
        )
    )
    .reset_index()
)

cost_analysis["cost_per_capacity_unit"] = (
    cost_analysis["total_cost"] /
    cost_analysis["available_capacity"]
)

cost_analysis["cost_per_demand_unit"] = (
    cost_analysis["total_cost"] /
    cost_analysis["demand_units"]
)

print("\n" + "=" * 60)
print("9. COST ANALYSIS")
print("=" * 60)

display(
    cost_analysis.sort_values(
        "total_cost",
        ascending=False
    )
)


# ============================================================
# 11. CURRENT VS PROJECTED ANALYSIS
# ============================================================

projection_summary = (
    df.groupby("region")
    .agg(
        current_demand=(
            "demand_units",
            "sum"
        ),
        projected_demand=(
            "projected_demand",
            "sum"
        ),
        current_capacity=(
            "available_capacity",
            "sum"
        ),
        projected_capacity=(
            "projected_capacity",
            "sum"
        )
    )
    .reset_index()
)

projection_summary["demand_growth_pct"] = np.where(
    projection_summary["current_demand"] > 0,
    (
        (
            projection_summary["projected_demand"] -
            projection_summary["current_demand"]
        )
        /
        projection_summary["current_demand"]
        * 100
    ),
    0
)

projection_summary["capacity_growth_pct"] = np.where(
    projection_summary["current_capacity"] > 0,
    (
        (
            projection_summary["projected_capacity"] -
            projection_summary["current_capacity"]
        )
        /
        projection_summary["current_capacity"]
        * 100
    ),
    0
)

print("\n" + "=" * 60)
print("10. CURRENT VS PROJECTED ANALYSIS")
print("=" * 60)

display(projection_summary)


# ============================================================
# 12. TOP REGIONS
# ============================================================

top_demand_region = (
    region_summary
    .sort_values(
        "demand_units",
        ascending=False
    )
    .head(5)
)

top_cost_region = (
    region_summary
    .sort_values(
        "total_cost",
        ascending=False
    )
    .head(5)
)

top_utilization_region = (
    region_summary
    .sort_values(
        "utilization_rate",
        ascending=False
    )
    .head(5)
)

print("\n" + "=" * 60)
print("11. TOP 5 REGIONS BY DEMAND")
print("=" * 60)

display(top_demand_region)


print("\n" + "=" * 60)
print("12. TOP 5 REGIONS BY COST")
print("=" * 60)

display(top_cost_region)


print("\n" + "=" * 60)
print("13. TOP 5 REGIONS BY UTILIZATION")
print("=" * 60)

display(top_utilization_region)


# ============================================================
# 13. FINAL BUSINESS INSIGHTS
# ============================================================

highest_demand_region = (
    region_summary
    .loc[
        region_summary["demand_units"].idxmax(),
        "region"
    ]
)

highest_cost_region = (
    region_summary
    .loc[
        region_summary["total_cost"].idxmax(),
        "region"
    ]
)

highest_utilization_region = (
    region_summary
    .loc[
        region_summary["utilization_rate"].idxmax(),
        "region"
    ]
)

print("\n" + "=" * 60)
print("14. FINAL BUSINESS INSIGHTS")
print("=" * 60)

print(
    "1. Highest demand region:",
    highest_demand_region
)

print(
    "2. Highest cost region:",
    highest_cost_region
)

print(
    "3. Highest utilization region:",
    highest_utilization_region
)

print(
    "4. Overall utilization: {:.2f}%"
    .format(overall_utilization)
)

print(
    "5. Projected utilization: {:.2f}%"
    .format(projected_utilization)
)

print(
    "6. Total cost:",
    round(total_cost, 2)
)

print(
    "7. Current shortage records:",
    len(shortage_df)
)

print(
    "8. Projected shortage records:",
    len(projected_shortage_df)
)


# ============================================================
# 14. CREATE ONE PYTHON ANALYSIS RESULT FILE
# ============================================================

kpi_result = pd.DataFrame({
    "Metric": [
        "Total Records",
        "Total Demand Units",
        "Total Available Capacity",
        "Overall Utilization (%)",
        "Total Projected Demand",
        "Total Projected Capacity",
        "Projected Utilization (%)",
        "Infrastructure Cost",
        "Operating Cost",
        "Total Cost",
        "Current Capacity Gap",
        "Projected Capacity Gap",
        "Current Shortage Records",
        "Projected Shortage Records"
    ],
    "Value": [
        total_records,
        total_demand,
        total_capacity,
        overall_utilization,
        total_projected_demand,
        total_projected_capacity,
        projected_utilization,
        total_infrastructure_cost,
        total_operating_cost,
        total_cost,
        capacity_gap,
        projected_gap,
        len(shortage_df),
        len(projected_shortage_df)
    ]
})

output_file = "Python Analysis Result.xlsx"

with pd.ExcelWriter(
    output_file,
    engine="openpyxl"
) as writer:

    kpi_result.to_excel(
        writer,
        sheet_name="Overall KPIs",
        index=False
    )

    region_summary.to_excel(
        writer,
        sheet_name="Region Analysis",
        index=False
    )

    segment_summary.to_excel(
        writer,
        sheet_name="Segment Analysis",
        index=False
    )

    service_summary.to_excel(
        writer,
        sheet_name="Service Analysis",
        index=False
    )

    monthly_summary.to_excel(
        writer,
        sheet_name="Monthly Analysis",
        index=False
    )

    if len(shortage_df) > 0:
        shortage_summary.to_excel(
            writer,
            sheet_name="Capacity Shortage",
            index=False
        )

    if len(projected_shortage_df) > 0:
        projected_shortage_summary.to_excel(
            writer,
            sheet_name="Projected Shortage",
            index=False
        )

    cost_analysis.to_excel(
        writer,
        sheet_name="Cost Analysis",
        index=False
    )

    projection_summary.to_excel(
        writer,
        sheet_name="Current vs Projected",
        index=False
    )


# ============================================================
# 15. COMPLETION MESSAGE
# ============================================================

print("\n" + "=" * 60)
print("PYTHON ANALYSIS COMPLETED SUCCESSFULLY")
print("=" * 60)

print(
    "\nAnalysis Result File Created:",
    output_file
)

print(
    "\nThis file contains all major analysis results in one Excel file."
)

print(
    "\nNext Step: Use the cleaned Excel file in Tableau for charts and dashboard."
)
```
